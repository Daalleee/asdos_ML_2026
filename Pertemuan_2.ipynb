```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

#### 1. Baca data

```python
df = pd.read_excel('data/gaji.xlsx')
df
```

##### 2. Isi field kosong dengan rata-rata (fillna)

```python
df["USIA"] = df["USIA"].fillna(df["USIA"].mean())
df["GAJI"] = df["GAJI"].fillna(df["GAJI"].mean())
df
```

##### 3. Scaler Z-Score untuk USIA dan GAJI

```python
scaler = StandardScaler()
df_zscore = scaler.fit_transform(df[["USIA", "GAJI"]])
df["Usia_ZScore"] = df_zscore[:, 0]
df["Gaji_ZScore"] = df_zscore[:, 1]
df[["USIA", "GAJI", "Usia_ZScore", "Gaji_ZScore"]]
```

#### 4. Normalisasi MinMax 0-1 untuk USIA dan GAJI

```python
minmax_01 = MinMaxScaler(feature_range=(0, 1))
df_minmax_01 = minmax_01.fit_transform(df[["USIA", "GAJI"]])
df["Usia_MinMax_01"] = df_minmax_01[:, 0]
df["Gaji_MinMax_01"] = df_minmax_01[:, 1]
df[["USIA", "GAJI", "Usia_MinMax_01", "Gaji_MinMax_01"]]
```

#### 5. Normalisasi MinMax 1-10 untuk USIA dan GAJI

```python
minmax_110 = MinMaxScaler(feature_range=(1, 10))
df_minmax_110 = minmax_110.fit_transform(df[["USIA", "GAJI"]])
df["Usia_MinMax_110"] = df_minmax_110[:, 0]
df["Gaji_MinMax_110"] = df_minmax_110[:, 1]
df[["USIA", "GAJI", "Usia_MinMax_110", "Gaji_MinMax_110"]]
```

#### 6. Gabungan dataframe asli + hasil proses

```python
df
```

#### 7. Export hasil.xlsx

```python
df.to_excel("Hasil.xlsx", index=False)
print("Berhasil")
```
